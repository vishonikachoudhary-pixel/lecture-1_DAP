```python
#Byjus case lect 2
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Lecture2_data - Lecture2_data.csv (1).csv")
```

```python
#decision making structures
#if else

revenue = df["Revenue_Lakh"].mean()

if revenue >= 300:
  print("Average revenue is high")
else:
  print("Average revenue is low")
```

```python
#decision making structure
# nested if statement in python

revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  print("Average revenue is high")
  if rating >= 4:
    print("Customer rating is good")
  else:
    print("Customer rating is bad needs improvement")
else:
  print("Average revenue is low")
```

```python
#find the average customer rating
rating = df["Customer_Rating"].mean()

#make a decision based on the rating
if rating >= 4.5:
  print("excellent")

elif rating >= 4.0:
  print("very good")
elif rating >= 3.0:
  print("average")

else:
  print("needs improvement")
```

```python
#for loop
for rating in df["Customer_Rating"]:

  if rating >= 4:
    print("good rating")
  else:
    print("Needs Information")
```

```python
#Byjus case lect 2
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Lecture2_data - Lecture2_data.csv (1).csv")
```

```python
#FOR loop with range() iloc = integer location

for i in range(10):
  print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#while loop in python
#while loop and i,j is variable to track

i = 0
while i < 5:
  revenue = df["Revenue_Lakh"].iloc[i]
  if revenue >= 300:
    print("revenue is high", revenue)
  else:
    print("revenue is low", revenue)
  i += 1
```
